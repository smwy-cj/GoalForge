# GoalForge 比对审计记录
本笔记本读取本轮保存的结果并复算计数。原始核验及语义判断保存在同目录脚本和记录中。
语义判断来自 Codex，不构成人工 Gold。工作目录为本笔记本目录或项目根目录。

```python
from pathlib import Path
import json
run_dir = Path.cwd()
if not (run_dir / 'metrics.json').exists():
    run_dir = run_dir / 'runs/codex-comparison-2026-10-22'
metrics = json.loads((run_dir / 'metrics.json').read_text(encoding='utf-8'))
print(json.dumps(metrics['counts'], ensure_ascii=False, indent=2))
```

```python
alignments = [json.loads(line) for line in (run_dir / 'goal-alignment.jsonl').read_text(encoding='utf-8').splitlines()]
from collections import Counter
print(dict(Counter(row['correspondence'] for row in alignments)))
print('Reasonable inference originally labeled explicit:', metrics['reasonable_inference_labeled_explicit_in_original'])
print('Paper-only goals pending full-source adjudication:', metrics['paper_only_source_not_assessed'])
```

```python
for split in ['calibration', 'consensus_evaluation', 'single_rater_evaluation']:
    result = metrics['relation_benchmark'][split]
    print(split, 'total=', result['n'], 'resolved=', result['resolved'], 'correct=', result['correct'], 'abstained=', result['abstained'])
print('Validation:', json.loads((run_dir / 'validation.json').read_text(encoding='utf-8'))['passed'])
```
